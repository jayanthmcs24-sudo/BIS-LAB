```python

import numpy as np

# Objective function
def fitness(x):
    return x**2 - 4*x + 4


# Initialize parameters
GRID_SIZE = 10
ITERATIONS = 100
LOWER_BOUND = -10
UPPER_BOUND = 10

# Initialize population randomly
cells = np.random.uniform(
    LOWER_BOUND,
    UPPER_BOUND,
    (GRID_SIZE, GRID_SIZE)
)

# Evaluate initial fitness
fitness_values = fitness(cells)

# Store global best solution
best_index = np.unravel_index(
    np.argmin(fitness_values),
    fitness_values.shape
)

best_x = cells[best_index]
best_fitness = fitness(best_x)


# Cellular algorithm
for iteration in range(ITERATIONS):

    new_cells = cells.copy()

    # Visit every cell
    for i in range(GRID_SIZE):
        for j in range(GRID_SIZE):

            # Get 3x3 neighborhood (excluding boundaries)
            neighbors = []

            for di in range(-1, 2):
                for dj in range(-1, 2):

                    ni = i + di
                    nj = j + dj

                    # Check valid grid position
                    if (0 <= ni < GRID_SIZE and
                        0 <= nj < GRID_SIZE):

                        neighbors.append(cells[ni, nj])

            # Find neighbor with minimum fitness
            neighbors = np.array(neighbors)

            best_neighbor = neighbors[
                np.argmin(fitness(neighbors))
            ]

            # Update cell using average
            new_cells[i, j] = (
                cells[i, j] + best_neighbor
            ) / 2

    # Replace old population
    cells = np.clip(
        new_cells,
        LOWER_BOUND,
        UPPER_BOUND
    )

    # Evaluate new fitness
    fitness_values = fitness(cells)

    # Find current best cell
    current_index = np.unravel_index(
        np.argmin(fitness_values),
        fitness_values.shape
    )

    current_best_x = cells[current_index]
    current_best_fitness = fitness(current_best_x)

    # Update global best
    if current_best_fitness < best_fitness:
        best_x = current_best_x
        best_fitness = current_best_fitness

# Display results
print("Parallel Cellular Algorithm Results")
print("------------------------------------")
print("Best value of x:", best_x)
print("Minimum fitness:", best_fitness)
print("Number of cells:", GRID_SIZE * GRID_SIZE)
print("Iterations:", ITERATIONS)
```

```python
import random

# Distance matrix between 4 cities
dist = [
    [0, 10, 15, 20],
    [10, 0, 35, 25],
    [15, 35, 0, 30],
    [20, 25, 30, 0]
]

n = len(dist)
ants = 10
iterations = 100
alpha, beta, rho, Q = 1, 2, 0.5, 100

pheromone = [[1.0] * n for _ in range(n)]
best_tour = None
best_length = float('inf')

for iteration in range(iterations):
    for ant in range(ants):
        tour = [random.randrange(n)]

        while len(tour) < n:
            current = tour[-1]
            unvisited = [c for c in range(n) if c not in tour]

            weights = [
                pheromone[current][c] ** alpha *
                (1 / dist[current][c]) ** beta
                for c in unvisited
            ]

            next_city = random.choices(
                unvisited, weights=weights
            )[0]
            tour.append(next_city)

        length = sum(
            dist[tour[i]][tour[(i + 1) % n]]
            for i in range(n)
        )

        if length < best_length:
            best_length = length
            best_tour = tour[:]

    # Evaporate pheromones
    for i in range(n):
        for j in range(n):
            pheromone[i][j] *= (1 - rho)

    # Deposit pheromones from the best tour
    for i in range(n):
        a = best_tour[i]
        b = best_tour[(i + 1) % n]
        pheromone[a][b] += Q / best_length
        pheromone[b][a] += Q / best_length

# Detailed output
route = best_tour + [best_tour[0]]

print("===== ACO FOR TSP =====")
print("Number of Cities:", n)
print("Number of Ants:", ants)
print("Number of Iterations:", iterations)
print("Best Tour:", " -> ".join(map(str, route)))
print("\nDistance Details:")

total = 0
for i in range(len(route) - 1):
    a, b = route[i], route[i + 1]
    d = dist[a][b]
    total += d
    print(f"City {a} -> City {b}: {d}")

print("\nShortest Total Distance:", total)
print("Starting City:", best_tour[0])
print("All Cities Visited:", len(set(best_tour)) == n)
print("Returns to Origin:", route[0] == route[-1])
```
